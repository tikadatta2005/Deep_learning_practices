```python
from torchvision import  transforms
from model1.model import CnnModel
import torch
from PIL import Image
```

```python
model = CnnModel().to("cuda")
model.load_state_dict(torch.load("./model1/model_epoch_30.pth"))
```

```python
transform = transforms.Compose([
    transforms.Resize((128, 128)),
    transforms.ToTensor()
])

img = Image.open("./data/test/dogs/dog.8643.jpg").convert("RGB")
img = transform(img)
```

```python
img
```

```python
img = img.unsqueeze(0)
img = img.to("cuda")
```

```python
model.eval()
with torch.no_grad():
    output = model(img)
    preds = torch.argmax(output, dim=1)
    if preds == 0: print("Cat")
    else: print("Dog")
```
