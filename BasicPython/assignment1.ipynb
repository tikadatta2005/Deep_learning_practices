# 1. Grocery Bill Calculator
- You are helping a local grocery store automate their billing system. Write a Python program that asks the user to enter the prices of three items and calculates the total bill. The program should also display the average price per item.

```python
prices = []

for i in range(3):
    price = int(input(f"Enter price of item {i+1}"))
    prices.append(price)

total = sum(prices)
avg = sum(prices)/3

print(f"Total Price: {total}")
print(f"Average Price: {avg}")
```

# 2. Student Grade Evaluator
- You are designing a simple grading system for a school. Write a Python program that asks the user to enter a student's score (0–100). Based on the score, print the grade using the following rules:

```python
marks = int(input("Enter mark of student"))
score_map = {90:"A", 80:"B", 70:"C", 60:"D", 0:"F"}

def get_grade():
    for key in score_map:
        if marks>=key: return score_map[key]

print(f"Grade ==> {get_grade()}")
```

# 3. Library Book Tracker
- You are building a system for a library to track borrowed books. Create a dictionary that stores book titles as keys and the number of days borrowed as values. Write a function that prints out all books borrowed for more than 7 days.

```python
books = {
    "Hello World":8,
    "Deep Learning": 15,
    "Machine Learning": 10,
    "Advance Mathematics": 2,
    "Rocket Science":1,
    "AI/ML 6 days Crash Course": 6
}

def print_books():
    for heading in books:
        if books[heading]>7: print(f"{heading}")

print_books()
```

# 4. Temperature Converter
- You are developing a tool for weather stations. Write a Python function that converts temperature from Celsius to Fahrenheit. The formula is:
* F = C × 9/5 + 32
- Ask the user to input a temperature in Celsius and display the result in Fahrenheit.

```python
def convert_to_fahrenheit(celcius): return celcius * (9/5) + 32

celcius = float(input("Enter Temperature in Celcius"))
print(f"Fahrenheit => {convert_to_fahrenheit(celcius)}")
```
